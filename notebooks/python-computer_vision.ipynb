```python
import cv2
import matplotlib.pyplot as plt

# Cargar imagen con OpenCV
img_bgr = cv2.imread("/content/chicken.webp")

# Convertir de BGR a RGB
img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)

# Redimensionar a 16x16 usando nearest neighbor
img_16 = cv2.resize(
    img_rgb,
    (16, 16),
    interpolation=cv2.INTER_NEAREST
)

# Mostrar imagen
plt.imshow(img_16)
plt.axis("off")
plt.show()
```
